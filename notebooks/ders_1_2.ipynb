```python
import sympy as sp

t = sp.symbols('t')                        # zaman sembolü
l, m = sp.symbols('l m', positive=True)    # sabit parametreler (pozitif dedik → sadeleştirme kolaylaşır)

x  = sp.Function('x')(t)                   # x(t): zamana bağlı bilinmeyen fonksiyon
th = sp.Function('theta')(t)               # θ(t)

xp = x + l*sp.sin(th)                      # uç: yatay konum
yp = l*sp.cos(th)                          # uç: yükseklik

vx = sp.diff(xp, t)                        # zamana göre türev → zincir kuralını sympy uygular
vy = sp.diff(yp, t)
print("ẋ_p =", vx)
print("ẏ_p =", vy)

v2 = sp.simplify(sp.expand(vx**2 + vy**2)) # hız² = ẋ_p² + ẏ_p², açıp sadeleştir
print("v² =", v2)
```

```python
M, g = sp.symbols('M g', positive=True)

xd = sp.diff(x, t)                                     # ẋ'yi bir değişkene koyduk, tekrar tekrar yazmamak için
T = sp.Rational(1, 2)*M*xd**2 + sp.Rational(1, 2)*m*v2 # Rational(1,2) = tam ½ (0.5 yazsak ondalık olurdu)
V = m*g*yp
L = T - V

p_x = sp.diff(L, xd)                                   # ∂L/∂ẋ  → x yönündeki "momentum"
print("∂L/∂ẋ =", sp.simplify(p_x))

EL_x = sp.diff(p_x, t) - sp.diff(L, x)                 # d/dt(∂L/∂ẋ) − ∂L/∂x
print("x denklemi:", sp.expand(EL_x), " = F")
```
